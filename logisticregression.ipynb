```python
#1. Why we need to move from linear to logistic regression:
#linear regression predicts contious values.
  #EX: House price prediction,Salary,Marks
  #But what if we want:
  Pass/Fail,Spam/Not Spam,Yes/No
  These are classification problems
  Linear Regression may predict:
  1.8
  -2.5
  0.8
  which does not make sense for classification problems
  so we need:
  output between 0 and 1
  That is why we use Logistic Regression for classification problems.
  
```

```python
#2.why it is called logistic regression
Logistic regression uses a regression equation internally:
z = mx + c
Then it applies a special function called:
Sigmoid Function:
z= 1/(1+e^(-z))
This converts values from:
-∞ to +∞
to
0 to 1
Hence it performs classification using regression mathematics.
```

```python
#sigmoid function calculation
import math
z = 2
probability = 1 / (1 + math.exp(-z))
print("z value:", z)
print("Sigmoid probability:", probability)

if probability >= 0.5:
    print("Class: Pass (Class 1)")
else:
    print("Class: Fail (Class 0)")
```

```python
#visualization of sigmoid curve
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-10, 10, 100)
y = 1 / (1 + np.exp(-x))
plt.figure(figsize=(10, 6))
plt.plot(x, y)
plt.xlabel('z values')
plt.ylabel('z ')
plt.title('Sigmoid Function')
plt.grid(True)
plt.show()
```
